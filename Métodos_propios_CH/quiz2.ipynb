```python

P = [(9, 0.5), (7, -3), (6, -1), (4, 2.5), (5, 12), (7, 15),
     (8, 12), (9.5, 13), (11, 15), (13, 13), (12, 10),
     (13, 3), (11, -4), (10, -1), (9, 0.5)]  

def calcular_area_orientada(puntos):
    n = len(puntos)
    area = 0
    for i in range(n):
        x_i, y_i = puntos[i]
        x_next, y_next = puntos[(i + 1) % n]  
        area += (x_i * y_next) - (x_next * y_i)
    return area / 2  
area = calcular_area_orientada(P)

if area < 0:
    P = P[::-1]  # Invertimos el orden de los puntos

print("Área orientada:", area)
print("Puntos organizados correctamente:")
for punto in P:
    print(punto)
```
